```python
from collections import defaultdict
import random

text = corpus.replace("\n", " ").strip()

state_size = 4
chain = defaultdict(list)

for i in range(len(text) - state_size):
    state = text[i:i + state_size]
    next_char = text[i + state_size]
    chain[state].append(next_char)

print("Character-level Markov chain created successfully!")
print("Number of states:", len(chain))
```

```python
def generate_text(chain, length=300):
    state = random.choice(list(chain.keys()))
    generated = state

    for _ in range(length):
        if state not in chain:
            break

        next_char = random.choice(chain[state])
        generated += next_char
        state = generated[-state_size:]

    return generated


sample = generate_text(chain, 300)

print(sample)
```

```python
import re
import random
from collections import defaultdict

sentences = [
    s.strip()
    for s in re.split(r'(?<=[.!?])\s+', corpus.strip())
    if s.strip()
]

print("Number of sentences:", len(sentences))
print("\nFirst 5 sentences:")
for i, sentence in enumerate(sentences[:5], 1):
    print(f"{i}. {sentence}")
```

```python
sentence_chain = defaultdict(list)

for i in range(len(sentences) - 1):
    current_sentence = sentences[i]
    next_sentence = sentences[i + 1]
    sentence_chain[current_sentence].append(next_sentence)

print("Sentence-level Markov chain created successfully!")
print("Number of states:", len(sentence_chain))
```
