```python
import tensorflow as tf
import keras_cv
from tensorflow import keras
import matplotlib.pyplot as plt

model = keras_cv.models.StableDiffusion(
    img_width=512,
    img_height=512
)
```

```python
images = model.text_to_image(
    "Realistic cinematic photograph of an astronaut riding a white horse, full horse visible, horse facing slightly toward the camera with a clear visible eye and detailed face, natural four-leg anatomy, astronaut fully visible, futuristic desert landscape at sunset, warm side lighting on the horse, balanced exposure, sharp focus, realistic textures, professional photography.",
    batch_size=1,
    num_steps=25
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 8))
plt.imshow(images[0])
plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
from PIL import Image

Image.fromarray(images[0]).save("task02_astronaut_horse.png")
print("Image saved successfully!")
```
